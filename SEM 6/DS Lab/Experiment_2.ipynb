```python
# data cleaning and preprocessing
df.replace('?', np.nan, inplace=True)
```

```python
df.dropna(axis=0)  
```

```python
df['max_life_span'] = pd.to_numeric(df['max_life_span'], errors='coerce')
```

```python
Q1 = df['max_life_span'].quantile(0.25)
Q3 = df['max_life_span'].quantile(0.75)
IQR = Q3 - Q1
df = df[~((df['max_life_span'] < (Q1 - 1.5 * IQR)) | (df['max_life_span'] > (Q3 + 1.5 * IQR)))]
df.dropna(axis=0, subset=['max_life_span'], inplace=True)  
print(df[["max_life_span"]])
```

```python
df.rename(columns={'total_sleep': 'total_sleep_hours'}, inplace=True)
print(df.columns)
```

```python
# data manipulation and transformation
df["brain_weight"] = pd.to_numeric(df["brain_weight"], errors="coerce")
df["body_weight"] = pd.to_numeric(df["body_weight"], errors="coerce")
df['total_weight'] = df['brain_weight'] + df['body_weight']
df["total_weight"] = pd.to_numeric(df["total_weight"], errors="coerce")
print(df[["total_weight"]])
```

```python
df['total_weight'] = df['total_weight'].fillna(df['total_weight'].mean())
print(df[["total_weight"]])
```

```python
# grouping and aggregation

print("1. Mean sleep hours by exposure index:")
print(df.groupby('sleep_exposure_index')['total_sleep_hours'].mean())
print()
print("2. Count of records by exposure index:")
print(df.groupby('sleep_exposure_index').size())
print()
print("3. Sleep statistics (mean, min, max) by exposure index:")
print(df.groupby('sleep_exposure_index')['total_sleep_hours'].agg(['mean', 'min', 'max']))
print()
print("4. Multiple aggregations across columns:")
print(df.groupby('sleep_exposure_index').agg({
    'total_sleep_hours': 'mean',
    'danger_index': 'mean',
    'total_weight': 'sum'
}))
print()
print("5. Describe statistics by exposure index:")
print(df.groupby('sleep_exposure_index')['total_weight'].describe())
```
